# AQI exploratory data analysis
Explore Kaggle's measured daily city-level pollutant observations and AQI labels. The source data does not contain temperature or humidity.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data_path = Path('../data/kaggle_city_day.csv')
df = pd.read_csv(data_path)
df.head()
```

```python
df.describe().T
```

```python
df.isna().mean().sort_values(ascending=False)
```

```python
features = ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3', 'Temperature', 'Humidity']
sns.heatmap(df[features + ['AQI']].corr(), cmap='coolwarm', center=0, annot=True, fmt='.2f')
plt.title('Feature and target correlations')
plt.tight_layout()
```

```python
sns.histplot(df['AQI'], bins=30, kde=True)
plt.title('Observed AQI distribution')
plt.tight_layout()
```
